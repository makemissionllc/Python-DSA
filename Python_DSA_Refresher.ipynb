```python

#Days to seconds
dys = input("Number of Days: ")
dys = float(dys)

ans = dys * 24 * 60 * 60

print(f"Answer: {int(ans)}")

```

```python

import math

radi = input("Enter the radius of the sphere: ")
radi = float(radi)

vol = (4 / 3) * math.pi * (radi**3)

print(f"The volume of a sphere with radius {radi} is: {vol}")

```

```python
def calculate_area(side):
    return side*side


def calculate_perimeter(side):
    return  side*4

sidlen = input("Enter the side length of the square: ")
sidlen = float(sidlen)

area = calculate_area(sidlen)
perimeter = calculate_perimeter(sidlen)

print(f"For a square with side length {sidlen}:")
print(f"Area: {area}")
print(f"Perimeter: {perimeter}")

```

```python
def cc(char):
    if char.isupper():
        return "uppercase"
    elif char.islower():
        return "lowercase"
    else:
        return "neither (it is not an alphabetic letter)"

usin = input("Enter a single character: ")

if len(usin) == 1:
    result = cc(usin)
    print(f"The character '{usin}' is {result}.")
else:
    print("Error: Please enter exactly one character.")
```

```python
x = 0
y = 20

while True:
    y -= 4 
    x += 2 / y

    if y < 6:
        break

print(f"Final Value: {round(x)}")
```

```python
val = []

for i in range(5):
    num = input(f"Enter value {i + 1} of 5: ")
    num = float(num)
    val.append(num)

ave = sum(val) / len(val)

# Display the entered values and the calculated average
print(f"\nValues entered: {val}")
print(f"Average of the values: {ave}")
```
